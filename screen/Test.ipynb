```python
import torch
from skimage import io
import os
#/scr/data/Micronuclei_data/screen_data/plate_1/20X_c0-DAPI-GFP

folder_path = "/scr/data/Micronuclei_data/screen_data/plate_1/20X_c0-DAPI-GFP"
output_path = "/scr/vidit/one_channel_plate_1"
#display all the filenames

file_names = os.listdir(folder_path)

def process_images_one_channel(folder_path, output_path, file_name):
    print(file_name)
    image = io.imread(folder_path + "/" + file_name)
    io.imsave(output_path + "/" + file_name, image[:, :, 1])
file_names
```

```python
#do it for all the filenames

for file_name in file_names:
    process_images_one_channel(folder_path=folder_path, output_path=output_path, file_name=file_name)
```
