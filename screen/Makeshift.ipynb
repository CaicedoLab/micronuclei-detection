```python
import os
import numpy as np
import zipfile
np.random.seed(42)

for i in range(8, 9):

    # Define the directory containing the images
    image_dir = f'/scr/vidit/screen-data/plate-{i}-channel-one'

    # Get a list of all image files in the directory
    all_files = os.listdir(image_dir)

    # Sample 1000 random files from the list of all files
    #sampled_files = np.random.choice(all_files, 1000, replace=False)

    # Define the path to the zip file
    zip_file_path = f'/scr/vidit/screen-data/plate_{i}_images.zip'

    # Create a zip file and add the sampled images to it
    with zipfile.ZipFile(zip_file_path, 'w') as zipf:
        for file in all_files:
            zipf.write(os.path.join(image_dir, file), file)

    # Print the path to the zip file
    print(f"Sampled images have been zipped and saved to: {zip_file_path}")
```
